#BEST STORE FOR PRODUCT

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
file_path="/content/sample_data/Apocolypse Food Prep.xlsx"
```

```python
df = pd.read_excel(file_path)

# View the first few rows
df.head()
```

```python
from pandas.core import groupby
df['Date']=pd.to_datetime(df['Date'])
print(df.shape, df.isna().sum())

basket =['Rice','Dried Beans', 'Bottled Water']
base = (df[df.Product.isin(basket)]
        .groupby(['Store','Product']).Price.mean()
        .groupby('Store').sum()
        .sort_values())
print(base)
```

```python
df['Date']=pd.to_datetime(df['Date'])
print(df.Date.min(), df.Date.max(), df.Date.nunique())
trend = (df[df.Product.isin(basket)]
               .groupby(['Store','Date'])['Price'].sum()
               .unstack())
print(trend)
trend.plot(marker='o', title='Basket cost over time')
plt.show()
```
